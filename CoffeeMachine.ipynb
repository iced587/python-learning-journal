```python
print('Welcome to the bean berets\nCOFFEE\nespresso, cappuccino, latte\nSIZES\nsmall, medium, large')

#order inputs
c_s = input('enter coffee size: ').lower()
c_t = input('enter coffee type: ').lower()

#validates size
def check_size(size):
  match size:
    case 'small' | 'medium' | 'large':
      return size
    case _:
      return 'Unknown Size'

#validates coffee type
def check_type(type_coffee):
  match type_coffee:
    case 'latte' | 'cappuccino' | 'espresso':
      return type_coffee
    case _:
      return 'Unknown Coffee'

#calls both other functions and finalizes order
def coffee_maker(size_input, type_input):
  final_size = check_size(size_input)
  final_type = check_type(type_input)
  if final_size == 'Unknown Size' or final_type == 'Unknown Coffee':
    return 'Enter Valid Order.'

#prints order
  print(f'Your order is a {final_size} {final_type}.')

#main function call
coffee_maker(size_input=c_s, type_input=c_t)
```
