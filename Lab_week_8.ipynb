Open the file. Prompt for the file name with input() and open it with open().
Check for errors. Loop over the file line by line, counting line numbers. Use strip() and skip blank lines and lines starting with #. Split each line with split(","), and skip any line that does not have 7 fields or has a score that is non-numeric or outside 0 to 100 (hint: isdigit()). Print a message for each skipped line.
Process grades. For each valid line, compute the weighted average and letter grade. Track the class average, highest, and lowest student.
Store the report in another file. Write the results to report.txt using with open("report.txt", "w"): one aligned row per student, then the summary.
Print the stored report. After the with block closes report.txt, open it again for reading and print every line to the console.

```python
import os
file_input = input('enter filename:')
file_name = file_input
with open(file_input, 'r') as file:

  highest_grade = ['','',0]

  students_graded = 0

  lines_skipped = 0

  lowest_grade = ['','',100]

  class_avg = 0

  weighted_list = []

  id_list = []

  last_names = []

  first_names = []

  letter_list = []

  for index, line in enumerate(file, start=1):
    valueSkip = False

    if line.strip() == '':
      print(f'skipping line {index}: empty line.')
      lines_skipped +=1
      continue
    elif (line.strip()).startswith('#'):
      print(f'skipping line {index}: starts with #.')
      lines_skipped +=1
      continue

    line = line.strip()
    fields = line.split(',')
    if len(fields) != 7:
      print(f'skipping line {index}: expected 7 fields, found {len(fields)}')
      lines_skipped +=1
      continue
    id, last, first, homework, quizzes, midterm, final = line.split(',')
    try:
      subTotal = int(homework) + int(quizzes) + int(midterm) + int(final)
    except ValueError:
      print(f'skippint line {index}: expected numeric score.')
      lines_skipped +=1
      continue
    for number in (homework,quizzes,midterm,final):
      if not number.isdigit or not 0 <= int(number) <=100:
        print(f'skipping line {index}: expected numeric score 1-100.')
        lines_skipped +=1
        valueSkip = True
        break
    if valueSkip == True:
      continue
    id_list.append(id)
    last_names.append(last)
    first_names.append(first)

    students_graded +=1

    weighted_homework = int(homework) * .20
    weighted_quizzes = int(quizzes) * .15
    weighted_midterm = int(midterm) * .25
    weighted_final = int(final) * .40

    weighted_grades = weighted_homework + weighted_quizzes + weighted_final + weighted_midterm
    weighted_list.append(weighted_grades)

    class_avg += weighted_grades

    if weighted_grades > highest_grade[2]:
      highest_grade = [last, first, weighted_grades]
    if weighted_grades < lowest_grade[2]:
      lowest_grade = [last, first, weighted_grades]
    letter_grade = ''
    if weighted_grades >= 90:
      letter_grade = 'A'
    elif weighted_grades >= 80:
      letter_grade = 'B'
    elif weighted_grades >= 70:
      letter_grade = 'C'
    elif weighted_grades >= 60:
      letter_grade = 'D'
    else:
      letter_grade = 'F'

    letter_list.append(letter_grade)
    lowest = f'{lowest_grade[0]},{lowest_grade[1]}  {lowest_grade[2]}'
    highest = f'{highest_grade[0]},{highest_grade[1]}  {highest_grade[2]}'

  if students_graded > 0:
    class_avg = class_avg / students_graded


report = open('report.txt', 'w')
report.write(f'''GRADE REPORT: {os.path.basename(file_name)}
ID\t\tName\t\t\t\t\t\t\tAverage\t\t\tGrade
-----------------------------------------\n''')
index = 0
while index < len(id_list):
  full_name = f'{last_names[index]},{first_names[index]}'
  report.write(f'{id_list[index]}\t{full_name}\t\t\t{weighted_list[index]}\t\t\t\t\t{letter_list[index]}\n')
  index+=1
report.write(f'''\nStudents Graded: {students_graded}
Lines Rejected: {lines_skipped}
Class Average: {class_avg}
Highest: {highest}
Lowest: {lowest}''')
report.close()

with open('report.txt', 'r') as file:
  final = file.read()
  print(final)

```
