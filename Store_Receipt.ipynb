```python
#subtotal variable
amounts = 0
#discount variable in percentage
discount_percent = 10

#returns price total by item and adds to subtotal
def item_total(price, quantity):
  amount = price * quantity
  global amounts
  amounts += amount
  return amount

#returns subtotal
def calculate_subtotal():
  return amounts

#returns total discount
def calculate_discount(subtotal, dis):
  global discount
  discount = (subtotal * dis)
  discount = discount / 100
  return discount

#returns final total
def calculate_final_total(subtotal, discount):
  return (subtotal - discount)

#all inputs
store_name = input('enter store name ')

product_one_name = input('enter product one\'s name ')
product_one_price = input('enter product one\'s price ')
product_one_quantity = input('enter product one\'s quantity ')
product_one_total = item_total(float(product_one_price),float(product_one_quantity))

product_two_name = input('enter product two\'s name ')
product_two_price = input('enter product two\'s price ')
product_two_quantity = input('enter product two\'s quantity ')
product_two_total = item_total(float(product_two_price),float(product_two_quantity))

product_three_name = input('enter product three\'s name ')
product_three_price = input('enter product three\'s price ')
product_three_quantity = input('enter product three\'s quantity ')
product_three_total = item_total(float(product_three_price),float(product_three_quantity))

discount_percent = input('enter discount percentage ')

final_discount = calculate_discount(amounts, float(discount_percent))

total = calculate_final_total(amounts,final_discount)

#Reciept printed
print(f'''{store_name}
{product_one_name} x{product_one_quantity}: ${product_one_total:.2f}
{product_two_name} x{product_two_quantity}: ${product_two_total:.2f}
{product_three_name} x{product_three_quantity}: ${product_three_total:.2f}
Subtotal: ${amounts:.2f}
Discount: ${final_discount:.2f}
Total: {total:.2f}$''')
```
