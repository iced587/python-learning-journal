```python
hoursList = []
hoursTotal = ()
dailyOvertime = 0

#Function to collect hours and stop invalid inputs
def collectHours():
  hoursFlag = False
  while not hoursFlag:
  #Input for hours of all five days
    try:
      hoursMon = int(input('Enter hours for Monday:'))
      hoursTue = int(input('Enter hours for Tuesday:'))
      hoursWed = int(input('Enter hours for Wednesday:'))
      hoursThr = int(input('Enter hours for Thursday:'))
      hoursFri = int(input('Enter hours for Friday:'))
  #Variables holding total hours and daily hours
      global hoursList
      hoursList = [hoursMon, hoursTue, hoursWed, hoursThr, hoursFri]
      global hoursTotal
      hoursTotal = (hoursMon + hoursTue + hoursWed + hoursThr + hoursFri)
  #Checks for invalid input
      if all(0 <= hours <= 24 for hours in (hoursMon, hoursTue, hoursWed, hoursThr, hoursFri)):
        hoursFlag = True
      else: print('Enter integers between 0-24.')
    except ValueError:
      print('Enter an integer between 0-24.')

#function to collect hourly wage
def collectWage():
  wageFlag = False
  while not wageFlag:
    try:
      #variable for users hourly wage
      global hourlyWage
      hourlyWage = int(input('Enter hourly wage. '))
      #a certain ceo's hourly wage (~3,500,000$)
      if (hourlyWage >= 3500000):
        print('↑→↓↓↓')
      #if your pay is 0$ or less contact your union rep
      elif (hourlyWage <= 0):
        print('Enter a positive integer above zero.')
      #passes function
      elif (0 < hourlyWage < 3500000):
        wageFlag = True
    #handles non integer inputs
    except ValueError:
      print('Enter a positive integer.')

#function to calculate total pay
def totalPay():
  collectWage()
  collectHours()
  global hoursTotal
  global dailyOvertime
  global hoursList

  #daily overtime calculation
  if hoursList[0] > 12:
    dailyOvertime += (hoursList[0] - 12)
    hoursList[0] = 12
  if hoursList[1] > 12:
    dailyOvertime += (hoursList[1] - 12)
    hoursList[1] = 12
  if hoursList[2] > 12:
    dailyOvertime += (hoursList[2] - 12)
    hoursList[2] = 12
  if hoursList[3] > 12:
    dailyOvertime += (hoursList[3] - 12)
    hoursList[3] = 12
  if hoursList[4] > 12:
    dailyOvertime += (hoursList[4] - 12)
    hoursList[4] = 12

#weekly overtime calculation
  if hoursTotal > 40:
    weekOvertime = hoursTotal - 40

#normal wage calculation
  wageMon = hoursList[0] * hourlyWage
  wageTue = hoursList[1]  * hourlyWage
  wageWed = hoursList[2]  * hourlyWage
  wageThr = hoursList[3]  * hourlyWage
  wageFri = hoursList[4]  * hourlyWage

  normalWages = (wageMon + wageTue + wageWed + wageThr + wageFri)

#all overtime pay calculation
  doubleOvertime = 0
  weekOverWages = 0
  allOverWages = 0
  doubleOverWages = 0
  while weekOvertime > 0 and dailyOvertime > 0:
    doubleOvertime += 1
    weekOvertime -= 1
    dailyOvertime -= 1
  while weekOvertime > 0:
    weekOverWages += (1.5 * hourlyWage)
    weekOvertime -=1
  while dailyOvertime > 0:
    allOverWages += (2 * hourlyWage)
    dailyOvertime -= 1
  while doubleOvertime > 0:
    doubleOverWages += (3 * hourlyWage)
    doubleOvertime -=1

  overtimeWages = (weekOverWages + allOverWages + doubleOverWages)

  totalWages = (overtimeWages + normalWages)

#print total pay
  print(f'This is your total pay for the week: {totalWages}')

#final function call
totalPay()
```
