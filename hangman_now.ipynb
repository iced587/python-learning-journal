```python
hangman = ['''
    ----+
    |   |
        |
        |
        |
==========
''','''
    ----+
    |   |
    o   |
        |
        |
==========
''','''
    ----+
    |   |
    o   |
    |   |
        |
==========
''','''
    ----+
    |   |
    o   |
   /|   |
        |
==========
'''
#raw string to stop the damn thing from imploding due to backslashes
,r'''
    ----+
    |   |
    o   |
   /|\  |
        |
==========
''',r'''
    ----+
    |   |
    o   |
   /|\  |
   /    |
==========
''',r'''
    ----+
    |   |
    o   |
   /|\  |
   / \  |
==========
'''
]


#create a list of words, take a random selection, store it as a list, display it as underscores.
import random
def initializeGame():
  words = ['hangman', 'justine', 'random', 'unique', 'crazy', 'fuel', 'input', 'hippopotomonstrosesquipedaliophobia']
  secretWord = list(random.choice(words))
  display = []
  for letter in secretWord:
    display += '_ '
  print(*display)
  return secretWord


def checkWin(display):
  if '_' not in display:
    return True

def updateDisplay(hiddenWord, guessed):
  display = []
  for letter in hiddenWord:
    if letter in guessed:
      display += letter + ' '
    else:
      display += "_ "
  return display


def userInput():
  print('Welcome to Hangman! Guess a letter: ')
  hiddenWord = initializeGame()
  guessed = []
  incorrectGuesses = 0


  while incorrectGuesses < 6:
    display = updateDisplay(hiddenWord, guessed)
    if checkWin(display) == True:
      print(*display)
      return('You Win!')


    print(hangman[incorrectGuesses])
    print(*display)
    uncheckedInput = input('Enter guess: ')
    if not uncheckedInput.isalpha():
      print('Enter a single letter')
      continue
    elif len(uncheckedInput) == 1:
      uInput = uncheckedInput
    else:
      print('Enter one letter.')
      continue
    if uInput not in guessed:
      guessed += uInput
    elif uInput in guessed:
      print('Already guessed.')
      continue
    if uInput not in hiddenWord:
      incorrectGuesses += 1
      print(f'Wrong! {incorrectGuesses} / 6')


  if incorrectGuesses == 6:
    print(hangman[6])
    return('You Lose :(')
#Take list of words and update the list of words

userInput()
```
